```python
zivoty = 3
pismena = []
inventar = ["Baterka"]
prohledana_mista = []
mista = [
    "Obývací pokoj",
    "Kuchyň",
    "Ložnice",
    "Sklep",
    "Pracovna"]


def uvod(inventar):
    print("Vítej ve hře: Emma se ztratila!")
    print("Jmenuješ se Honza a hledáš svou přítelkyni Emmu.")
    print("Už jsi ji 9 dní neviděl. Naposledy byla ve starém domě, do kterého jsi teď vešel.")
    print("Musíš prozkoumat dům a najít heslo k truhle.")
    print("Máš 3 životy.")
    
    print("Na začátku máš v inventáři:")
    print(inventar)


def pridej_predmet(predmet, inventar):
        if predmet not in inventar:
            inventar.append(predmet)
            print("Získal jsi:", predmet)
        else:
            print("Tento předmět už máš.")
        return inventar


def pouzij_predmet(inventar, zivoty):
    if "Lékárnička" not in inventar or zivoty == 3:
        return inventar, zivoty
    print("Máš lékárničku.")
    while True:
        volba = input("Chceš použít lékárničku? ano/ne: ").strip().lower()
        if volba == "ano":
            zivoty = zivoty + 1
            inventar.remove("Lékárnička")
            print("Získal jsi jeden život.")
            print("Lékárnička byla spotřebována.")
            break

        elif volba == "ne":
            print("Lékárničku si zatím necháš.")
            break

        else:
            print("Zadej pouze ano nebo ne.")
    return inventar, zivoty


def prozkoumej(misto, pismena, inventar, zivoty):
    prozkoumano = True
    if misto == "Obývací pokoj":
        print("Na stole najdeš malý papírek s písmenem D.")
        if "D" not in pismena:
            pismena.append("D")
            print("Našel jsi písmeno D.")

        print("Pod papírkem leží malý klíč.")
        inventar = pridej_predmet("Klíč", inventar)

    elif misto == "Kuchyň":
        print("Na kuchyňském stole je vysypaná sůl ve tvaru U.")
        if "U" not in pismena:
            pismena.append("U")
            print("Našel jsi písmeno U.")

        print("Na zemi leží lékárnička.")
        inventar = pridej_predmet("Lékárnička", inventar)
    
    elif misto == "Ložnice":
        if "Klíč" in inventar:
            if "Baterka" in inventar:
                while True:
                    odpoved = input("Chceš použít baterku? ano/ne: ").strip().lower()
                    if odpoved == "ano" or odpoved == "ne":
                        break
                    else:
                        print("Zadej pouze ano nebo ne.")
                print("V ložnici je zamčená skříň, ale pomocí klíče ji otevřeš.")
                if odpoved == "ano":
                    print("Díky baterce vidíš dovnitř skříně.")
                    inventar.remove("Baterka")
                    print("Baterka se vybila a už ji nemáš.")
                    print("Uvnitř najdeš dřevěné písmeno M.")
                    if "M" not in pismena:
                        pismena.append("M")
                        print("Našel jsi písmeno M.")
                        
                    inventar.remove("Klíč")
                    print("Klíč jsi použil a už ho nemáš.")
                    
                else:
                    print("Bez baterky nevidíš dovnitř skříně.")
                    print("Musíš hledat dál.")
                    prozkoumano = False
                    
            else:
                print("Nemáš baterku, takže ve skříni nic nevidíš.")
                print("Musíš hledat dál.")

        else:
            print("V ložnici je jenom skříň, ale je zamčená. Potřebuješ klíč.")
            print("Můžeš se sem vrátit později.")
            prozkoumano = False

    elif misto == "Sklep":
        if "Baterka" in inventar:
            while True:
                odpoved = input("Chceš použít baterku? ano/ne: ").strip().lower()
                if odpoved == "ano":
                    print("Díky baterce vidíš do tmavého sklepa.")
                    inventar.remove("Baterka")
                    print("Baterka se vybila a už ji nemáš.")
                    break

                elif odpoved == "ne":
                    print("Ve sklepě je tma a zakopl jsi.")
                    zivoty = zivoty - 1
                    print("Ztratil jsi jeden život.")
                    break

                else:
                    print("Zadej pouze ano nebo ne.")

        else:
            print("Nemáš baterku. Ve sklepě je tma a zakopneš.")
            zivoty = zivoty - 1
            print("Ztratil jsi jeden život.")
        print("Ve sklepě jsi nic nenašel. Můžeš hledat dál.")

    elif misto == "Pracovna":
        print("Vstoupil jsi do pracovny.")
        print("Je tam zapnutý počítač a svítí na něm písmeno.")
        if "M" not in pismena:
            pismena.append("M")
            print("Našel jsi písmeno M.")
    return pismena, inventar, zivoty, prozkoumano


def truhla(zivoty):
    pokusy = 0
    vyhra = False
    print("Našel jsi starou truhlu.")
    print("Zadej heslo, které jsi našel.")
    while zivoty > 0:
        heslo = input("Heslo: ").strip().upper()
        if heslo == "DUM" or heslo == "DŮM":
            print("Truhla se otevřela!")
            print("Uvnitř je klíč, který otevře skrytou místnost a tam najdeš Emmu.")
            print('"Honzo! Zachránil jsi mě!" řekne Emma.')
            print("KONEC HRY")
            vyhra = True
            break

        else:
            print("Špatné heslo.")
            pokusy = pokusy + 1
            if pokusy % 2 == 0:
                zivoty = zivoty - 1
                print("Ztratil jsi jeden život.")
                print("Zbývá ti životů:", zivoty)
            if pokusy == 3:
                print("Nápověda: Heslo je spojené s domem.")
    if zivoty == 0:
        print("Došly ti životy. Prohrál jsi.")
    return zivoty, vyhra

uvod(inventar)

vyhra = False

while (
    zivoty > 0
    and len(prohledana_mista) < len(mista)
    and not ("D" in pismena and "U" in pismena and "M" in pismena)
    and vyhra == False):
    
    print("Máš životy:", zivoty)
    print("Nalezená písmena:", pismena)
    print("Kam chceš jít?")
    
    for i in range(len(mista)):
        if mista[i] not in prohledana_mista:
            print(i + 1, "-", mista[i])

    try:
        volba = int(input("Vyber místnost: "))
        if volba > 0 and volba <= len(mista):
            misto = mista[volba - 1]

            if misto not in prohledana_mista:
                pismena, inventar, zivoty, prozkoumano = prozkoumej (misto, pismena, inventar, zivoty)

                if prozkoumano:
                    prohledana_mista.append(misto)

                if prozkoumano and zivoty > 0:
                    inventar, zivoty = pouzij_predmet(inventar, zivoty)

            else:
                print("Tuto místnost už jsi prohledal.")

        else:
            print("Taková místnost neexistuje.")

    except ValueError:
        print("Musíš zadat číslo.")


if zivoty > 0:
    if "D" in pismena and "U" in pismena and "M" in pismena:
        print("Našel jsi všechna tři písmena!")
        zivoty, vyhra = truhla(zivoty)
    else:
        print("Nenašel jsi všechna písmena. Truhla zůstane zamčená.")
```
